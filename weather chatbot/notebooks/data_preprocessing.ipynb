```python
#Step 6: Text Cleaning and Preprocessing
"""What is the Temperature?
what is the temperature
WHAT IS THE TEMPERATURE!!
What's the temperature? To a computer, these look like different sentences. So we normalize them into a consistent format.
 org text ---lowercase ----remove punctuation -----tokenization ------join words again------clean sentence 
 re.sub(pattern, replacement, string) text = "Hello!!!"   result = re.sub("!", "", text) so o/p Hello
 By default, WordNetLemmatizer assumes every word is a noun (pos='n') """
import re
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Create lemmatizer object
lemmatizer = WordNetLemmatizer()


def clean_text(text):

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z0-9\s]", "", text)

    # Tokenize
    words = word_tokenize(text)

    # Lemmatize every word
    lemmatized_words = []

    for word in words:
        lemmatized_words.append(lemmatizer.lemmatize(word))

    words = lemmatized_words

    # Join words back into a sentence
    text = " ".join(words)

    return text
#a="What's writing?"
#print(clean_text(a))
```

```python
"""Apply Cleaning on every sentence(pattern) in the DataFram """
df["pattern"] = df["pattern"].apply(clean_text)
```

```python
""" End of Step 6

At this point, your data is clean, normalized, and ready for numerical conversion.

The next step (Step 7) is to:
Convert the cleaned text into integer sequences using Keras Tokenizer.
Build the vocabulary (word_index).
Understand how words are mapped to numbers.
Prepare the sequences that will later be padded and fed into the LSTM.

This is the step where the model starts seeing numbers instead of text.
Sentence
↓
Current temperature today
↓
Split into words
↓
Current
Temperature
Today
↓
Assign IDs
↓
3
4
5
↓
Sequence
↓

[3,4,5]

step 7.1 Create Tokenizer , is an empty object. It knows nothing about your vocabulary.
Step 7.2 Learn Vocabulary
Now train the tokenizer on your dataset. What does fit_on_texts() do?
It scans every sentence.
Suppose your dataset is "hi nice"
tokenizer reades hi -> 1 nice
example
{
'hi':1,
'hello':2,
'current':3,
'temperature':4,
'humidity':5
} This dictionary is called the Word Index.


the tokenizer learns your entire vocabulary.
{
    'current': 1,
    'temperature': 2,
    'humidity': 3,
    'wind': 4,
    'speed': 5,
    'hi': 6,
    'hello': 7 ......
}This dictionary is stored inside the tokenizer.
  
  *****  word_index
         Maps each word to an integer.
         {
'temperature':1,
'current':2,
'humidity':3
}

*******  word_counts
Counts how many times each word appeared.
OrderedDict([
('temperature', 42),
('current', 39),
('humidity', 21)
])

 *******document_count
How many training sentences were seen.

*********
What does pickle.dump() save?
When you do pickle.dump(tokenizer, file)
Python saves the entire tokenizer object.

That means it saves:
Tokenizer
│
├── word_index
├── word_counts
├── word_docs
├── document_count
├── configuration
└── everything needed later

It does not save:

❌ sequences
❌ DataFrame
❌ intents.json
❌ responses

Only the tokenizer.
"""
```

```python
# ============================================================
# Step 7: Convert Text into Integer Sequences
# ============================================================

# Import Tokenizer
from tensorflow.keras.preprocessing.text import Tokenizer

# Create Tokenizer object
tokenizer = Tokenizer()

# Learn vocabulary from all patterns
tokenizer.fit_on_texts(df["pattern"])

# Display the vocabulary (word -> index)
print("Word Index:\n")
print(tokenizer.word_index)

# Display vocabulary size
print("\nVocabulary Size:", len(tokenizer.word_index))

# Convert each sentence into a sequence of integers
sequences = tokenizer.texts_to_sequences(df["pattern"])

# Display first 10 sentences and their sequences
print("\nFirst 10 Sequences:\n")

for i in range(10):
    print(f"Sentence {i+1}: {df['pattern'][i]}")
    print(f"Sequence  : {sequences[i]}")
    print("-" * 50)

# Example: Convert a new sentence into sequence
test_sentence = ["current temperature today"]

test_sequence = tokenizer.texts_to_sequences(test_sentence)

print("\nExample Sentence:")
print(test_sentence[0])

print("Converted Sequence:")
print(test_sequence)

# Save tokenizer for future prediction
import pickle

with open("../models/tokenizer.pkl", "wb") as file:
    pickle.dump(tokenizer, file)

print("\nTokenizer saved successfully!")
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Find the maximum sentence length
max_length = max(len(sequence) for sequence in sequences)

print("Maximum Sentence Length:", max_length)

# Pad all sequences
padded_sequences = pad_sequences(
    sequences,
    maxlen=max_length,
    padding="post"
)

# Check shape
print("Shape:", padded_sequences.shape)

# View first 5 examples
for i in range(5):
    print("Original :", sequences[i])
    print("Padded   :", padded_sequences[i])
    print("--------------------------------------------------------") # use this both same print("-" * 50)
```

```python
"""Until now, your chatbot knows the questions (patterns), but the model still doesn't understand the intent names because they are text.

For example:

Pattern	Tag
Current temperature	current_temperature
Humidity today	humidity
Wind speed	wind_speed

An LSTM cannot predict text labels. It can only predict numbers.

So we convert:

current_temperature → 0
humidity            → 1
wind_speed          → 2
...

This process is called Label Encoding."""
```

```python
# ============================================================
# Step 9: Label Encoding
# ============================================================

from sklearn.preprocessing import LabelEncoder
import pickle

# Create LabelEncoder
label_encoder = LabelEncoder()

# Learn all intent classes
label_encoder.fit(df["tag"])

# Convert text labels into numbers
encoded_labels = label_encoder.transform(df["tag"])

# Display first 10 encoded labels
print("First 10 Encoded Labels:\n")

for i in range(20):
    print(f"Tag      : {df['tag'][i]}")
    print(f"Encoded  : {encoded_labels[i]}")
    print("-" * 40)

# Display all classes
print("\nIntent Classes:")
print(label_encoder.classes_)

# Display mapping
print("\nLabel Mapping:")

for index, label in enumerate(label_encoder.classes_):
    print(f"{index} -> {label}")

# Save LabelEncoder
with open("../models/label_encoder.pkl", "wb") as file:
    pickle.dump(label_encoder, file)

print("\nLabel Encoder saved successfully!")
```

```python
""" Now your dataset is completely preprocessed.
✅ padded_sequences → Input (X)
✅ encoded_labels → Output (y)

Now we divide the data into training and testing sets
Step 10: Train-Test Split
Goal
Split the dataset into:

Training Data (80%) → Used to train the LSTM.
Testing Data (20%) → Used to evaluate the LSTM on unseen data.

Without stratify
The split is random.
Suppose
Training

Greeting = 43
Weather = 32
Goodbye = 5

Testing

Greeting = 7
Weather = 8
Goodbye = 5

Notice something?
Half of all Goodbye samples ended up in the test set.
The model only learns from 5 Goodbye examples.
This is not ideal.
Even worse, sometimes a rare class may end up entirely in the test set, leaving the model with no examples to learn from.

With stratify=encoded_labels
The function looks at the class labels and tries to maintain the same proportions in both training and testing sets.
Original data:

Label	Percentage
Greeting	50%
Weather	    40%
Goodbye	    10%

If you split 80% train and 20% test:

Training (80 samples)

Label	Count
Greeting	40
Weather  	32
Goodbye	     8

Testing (20 samples)

Label	Count
Greeting	10
Weather	     8
Goodbye	     2

Notice the percentages stay the same.
stratify=encoded_labels
you're telling train_test_split:
"While splitting the data, keep the distribution of these labels the same in the training and testing sets."
"""
```

```python
# ============================================================
# Step 10: Train-Test Split
# ============================================================

from sklearn.model_selection import train_test_split

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    padded_sequences,
    encoded_labels,
    test_size=0.2,
    random_state=42,
    stratify=encoded_labels
)

# Display shapes
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)

print()

print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)

# Display one sample
print("\nFirst Training Sequence:")
print(X_train[0])

print("\nEncoded Label:")
print(y_train[0])

print("\nOriginal Intent:")
print(label_encoder.inverse_transform([y_train[0]]))
```
